```python
# Task 11: Create a Delivery Performance Function
def check_delivery_performance(Delivery_status, Days_since_order, Expected_delivery_days):
    if Delivery_status == "Cancelled":
        return "Cancelled"
    elif Delivery_status == "Delivered":
        return "Completed"
    elif Days_since_order <= Expected_delivery_days:
        return  "On Time"
    else:
        return "Delayed"

delivery_performance = check_delivery_performance("In Transit", 6, 5)
print (delivery_performance)
```

```python
# Task 12: Create a Priority Function
def determine_delivery_priority(Delivery_performance, Days_since_order, Expected_delivery_days, Delivery_attempts):
    if Delivery_Performance == "Delayed" and Delivery_attempts >= 2:
        return "Urgent"
    elif Delivery_Performance == "Delayed" and Delivery_attempts >= 1:
        return "High"
    elif Delivery_Performance == "On Time" and Delivery_attempts >= 1:
        return "Normal"
    else:
        return "Low"


delivery_priority = determine_delivery_priority("Delayed", 4, 5, 1)
print(delivery_priority)
```

```python
# Task 13: Create an Action Function
def recommend_delivery_action(Priority):
    if Priority == "Urgent":
        return "Contact Customer and Escalate Delivery."
    elif Priority == "High":
        return "Investigate Delivery Delay."
    elif Priority == "High":
        return "Continue Delivery Process."    
    else:
        return "Continue Delivery Process."

Recommended_action = recommend_delivery_action("High")
print (Recommended_action)
```

```python
# Task 14: Create a Delivery Validation Function
def validate_delivery(Delivery_ID_Validation, Customer_name_Validation,Delivery_location_Validation,
                      Delivery_status_Validation,Days_since_order_Validation,Expected_delivery_days_Validation,
                      Package_weight_Validation,Delivery_attempts_Validation ):
    if (Delivery_ID_Validation and Customer_name_Validation and Delivery_location_Validation and
                      Delivery_status_Validation and Days_since_order_Validation and Expected_delivery_days_Validation and
                      Package_weight_Validation and Delivery_attempts_Validation):
        return "VALID"
    else:
        return "INVALID"

Validation = validate_delivery("DEL001", "Daniel Okafor", "Lagos", "Processing", 4, 5, 3.5, 1)
print(Validation)
```
